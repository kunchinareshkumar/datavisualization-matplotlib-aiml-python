```python

# Pie chart variagle useful for Dept instead of numerical values

# Add new column as well for next visuzalise
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# 1D dictionary 
sal = {"Salary":[1000,2000,4000,5000,7000,60000,40000,100000]}

df = pd.DataFrame(sal)
df["dept"]=["HR","IT","Finance","HR","FE","backend","DEV","backend"]
df.info()
```

```python


count =df["dept"].value_counts()
plt.pie(count)  #without labels
plt.show()

plt.pie(count,labels=count.index, autopct="%1.0f")  #with labels,labels and autipct
plt.show()


# Explode


plt.pie(count,labels=count.index, autopct="%1.0f", explode = [0.1, 0, 0, 0,0.1,0.5])  #with labels,labels and autipct
plt.axis("equal") # shows piechart in eqal
plt.show()







```

```python
# Dept in Bar charts

count =df["dept"].value_counts()
plt.bar(count.index, count)  #without labels
plt.show()

# variaus colors
plt.bar(count.index, count, color="red")  #without labels


# variaus colors
plt.bar(count.index, count, color=["red","blue","green","black"])  #without labels

```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
df = pd.read_csv("customers.csv")
df.head(10)
```

```python
# Table data has labels data/category data instead if numnerical so we need this below approch

customerId =df["Customer Id"].value_counts()
plt.figure(figsize=(10, 5))
customerId.head(10).plot(kind="bar", color=["green", "red"])
plt.title("Top Cities Distribution")
plt.xlabel("city")
plt.ylabel("Count")

firstname=df["First Name"].value_counts()
plt.figure(figsize=(10, 5))
firstname.head(10).plot(kind="bar",color="blue")
plt.grid()

lastname=df["Last Name"].value_counts()
plt.grid()
plt.figure(figsize=(10, 5))
lastname.head(10).plot(kind="bar",color="red")


company=df["Company"].value_counts()
plt.grid()
plt.figure(figsize=(10,10))
company.head(10).plot(kind="bar",color="orange")


city=df["City"].value_counts()
plt.grid()
plt.figure(figsize=(10,10))
city.head(10).plot(kind="bar",color="black")


email=df["Email"].value_counts()
plt.grid()
plt.figure(figsize=(10,10))
email.head(10).plot(kind="bar",color="yellow")






```

```python
# Table data has labels data/category data instead if numnerical so we need this below approch

customerId =df["Customer Id"].value_counts()

customerId.head(10).plot(kind="pie",labels=customerId.index, autopct="%1.0f")
plt.title("Top Cities Distribution")
plt.xlabel("city")
plt.ylabel("Count")

firstname=df["First Name"].value_counts()
plt.figure()
firstname.head(10).plot(kind="pie",labels=firstname.index, autopct="%1.0f")
plt.grid()

lastname=df["Last Name"].value_counts()
plt.grid()
plt.figure()
lastname.head(10).plot(kind="pie",labels=lastname.index, autopct="%1.0f")


company=df["Company"].value_counts()
plt.grid()
plt.figure()
company.head(10).plot(kind="pie",labels=company.index, autopct="%1.0f")


city=df["City"].value_counts()
plt.grid()
plt.figure()
city.head(10).plot(kind="pie",labels=city.index, autopct="%1.0f")


email=df["Email"].value_counts()
plt.grid()
plt.figure()
email.head(10).plot(kind="pie",labels=email.index, autopct="%1.0f")
```
